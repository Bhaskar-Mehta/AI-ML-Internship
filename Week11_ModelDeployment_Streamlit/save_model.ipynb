# Week 11 – Model Deployment Basics
### Day 1: Train and Save the Model

Before an app can use a model, the trained weights have to be saved to a file. Here I retrain the Week 10 CNN and save its weights as `mnist_cnn.pt`.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from model import SimpleCNN

transform = transforms.ToTensor()
train_data = datasets.MNIST(root="./data", train=True, download=True, transform=transform)
test_data = datasets.MNIST(root="./data", train=False, download=True, transform=transform)
train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)

model = SimpleCNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

for epoch in range(5):
    running_loss = 0
    for images, labels in train_loader:
        optimizer.zero_grad()
        loss = criterion(model(images), labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}/5, Loss: {running_loss/len(train_loader):.4f}")
```

```python
torch.save(model.state_dict(), "mnist_cnn.pt")
print("Model saved to mnist_cnn.pt")
```

### Day 2: Reload the Saved Model

A saved model is only useful if loading it gives the same results. I load it into a fresh model and check the accuracy matches Week 10.

```python
loaded_model = SimpleCNN()
loaded_model.load_state_dict(torch.load("mnist_cnn.pt", map_location="cpu"))
loaded_model.eval()

correct, total = 0, 0
with torch.no_grad():
    for images, labels in test_loader:
        _, predicted = torch.max(loaded_model(images), 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Reloaded model test accuracy: {correct/total:.4f}")
```
