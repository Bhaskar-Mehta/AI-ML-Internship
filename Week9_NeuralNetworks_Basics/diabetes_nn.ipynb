# Week 9 – Neural Networks: The Basics
### Day 1: How Neural Networks Work (Concepts)

Before writing any code, here's a plain-language summary of what a neural network actually is.

**Neuron:** A tiny function that takes inputs, multiplies each by a "weight," adds them up, adds a bias, and passes the result through an activation function (like ReLU or Sigmoid) to decide its output.

**Layer:** A group of neurons that all process the same inputs in parallel. A network is usually built from an input layer, one or more hidden layers, and an output layer.

**Weights:** Numbers that control how much influence each input has. These start out random and get adjusted during training.

**Forward pass:** Data flows through the network layer by layer to produce a prediction.

**Loss function:** Measures how wrong the prediction was compared to the actual answer.

**Backpropagation:** The network works backward from the loss, calculating how much each weight contributed to the error.

**Gradient descent:** Weights are nudged slightly in the direction that reduces the error, a little bit at a time, over many rounds (epochs).

In short: a neural network is a stack of simple weighted calculations that gradually adjusts itself to get better at mapping inputs to outputs.
